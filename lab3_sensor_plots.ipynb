```python
# Generate simulated temperature readings from two sensors.
# Sensor A: mean 25 C, std 3 C, 200 readings.
# Sensor B: mean 27 C, std 4.5 C, 200 readings.
# Also generate 200 timestamps uniformly from 0 to 10 seconds.
# Use np.random.default_rng with a seed = last 4 digits of your Drexel ID.

import numpy as np
import matplotlib.pyplot as plt

# Set seed to last 4 digits of Drexel ID (replace with your actual ID)
seed = 1234  # Example seed; replace with your Drexel ID's last 4 digits

# Create random number generator
rng = np.random.default_rng(seed)

# Generate sensor data
sensor_a = rng.normal(25, 3, 200)
sensor_b = rng.normal(27, 4.5, 200)

# Generate timestamps: evenly spaced from 0 to 10 seconds (200 points)
timestamps = np.linspace(0, 10, 200)
```

```python
# Scatter plot: sensor reading (y-axis) vs timestamp in seconds (x-axis).
# Color Sensor A points blue and Sensor B points orange.
# Add axis labels with units, a title, and a legend.

plt.figure(figsize=(10, 6))
plt.scatter(timestamps, sensor_a, color='blue', label='Sensor A', alpha=0.7)
plt.scatter(timestamps, sensor_b, color='orange', label='Sensor B', alpha=0.7)
plt.xlabel('Time (seconds)')
plt.ylabel('Temperature (°C)')
plt.title('Sensor Temperature Readings Over Time')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# Overlaid histogram of Sensor A and Sensor B temperature distributions.
# Use 30 bins, alpha=0.5 for transparency so both distributions are visible.
# Add vertical dashed lines at each sensor's mean.
# Include a legend labeling each sensor.

plt.figure(figsize=(10, 6))
plt.hist(sensor_a, bins=30, alpha=0.5, color='blue', label='Sensor A')
plt.hist(sensor_b, bins=30, alpha=0.5, color='orange', label='Sensor B')
plt.axvline(np.mean(sensor_a), color='blue', linestyle='--', linewidth=2, label=f'Sensor A Mean: {np.mean(sensor_a):.2f}°C')
plt.axvline(np.mean(sensor_b), color='orange', linestyle='--', linewidth=2, label=f'Sensor B Mean: {np.mean(sensor_b):.2f}°C')
plt.xlabel('Temperature (°C)')
plt.ylabel('Frequency')
plt.title('Temperature Distributions for Sensor A and Sensor B')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```
